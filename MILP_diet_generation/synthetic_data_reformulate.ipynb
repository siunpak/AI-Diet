```python
### 일주일 식단 작성 ###

def month_diet(db, db_mor):
    
    month_diet = pd.DataFrame()
    for i in range(4):
        week_diet = []
        for i in range(7):
            if i < 4:
                week_diet.append(db_mor.loc[np.random.choice(db_mor.index)])
            else:
                week_diet.append(db.loc[np.random.choice(db.index)])

        month_diet = pd.concat([month_diet, pd.DataFrame(week_diet)])


    return month_diet.reset_index(drop=True)
```
